**Question #:1**

**Which SQL code snippet will correctly demonstrate a Data Definition Language (DDL) operation used to create a table?**

A. CREATE TABLE employees ( id INT,
   name STRING
   );

B. DROP TABLE employees;

C. ALTER TABLE employees ADD COLUMN salary DECIMAL(10,2);

D. INSERT INTO employees (id, name) VALUES (1 'Alice');


**The correct answer is:**

---

✅ **A. CREATE TABLE employees ( id INT, name STRING );**

---

**Explanation**

DDL (Data Definition Language) is used to define or modify the structure of database objects such as tables, schemas, and indexes.

---

- ✅ **A. CREATE TABLE employees ( id INT, name STRING );**  
  Creates a new table named employees.  
  This is a DDL operation.

- ❌ **B. DROP TABLE employees;**  
  Also a DDL operation because it removes a table.  
  However, it does not create a table, so it doesn't answer the question.

- ❌ **C. ALTER TABLE employees ADD COLUMN salary DECIMAL(10,2);**  
  Also a DDL operation because it modifies the table structure.  
  It alters an existing table rather than creating one.

- ❌ **D. INSERT INTO employees (id, name) VALUES (1, 'Alice');**  
  This is DML (Data Manipulation Language) because it inserts data into a table.  
  (Note: The option as written is also missing a comma after 1; the correct syntax is VALUES (1, 'Alice').)

**Question #:9**

**A data engineer is developing an ETL process based on Spark SQL. The execution fails. The data engineer checks the Spark UI and can see the ERRORS as follows: `"java.lang.OutofMemoryError: Java heap space"`**

**Which two corrective actions should the data engineer perform to resolve this issue? (Choose two.)**

A. Narrow the filters in order to collect less data in the query  
B. Upsize the worker nodes and activate autoshuffle partitions  
C. Upsize the driver node and deactivate autoshuffle partitions  
D. Cache the dataset in order to boost the query performance  
E. Fix the shuffle partitions to 50 to ensure the allocation

---
**The correct answers are: ✅ A and ✅ B**

---

**Why?**

The error:


java.lang.OutOfMemoryError: Java heap space


means that Spark does not have enough JVM heap memory to process the workload. This is commonly caused by:

- Processing too much data at once
- Large shuffle operations
- Insufficient executor (worker) memory
- Data skew or inefficient partitioning

---
##### ✅ A. Narrow the filters in order to collect less data in the query

**Correct.**

Reducing the amount of data processed lowers memory usage.

**Examples:**

- Apply filters as early as possible.
- Read only required columns.
- Use partition pruning.

This reduces the data Spark must keep in memory and shuffle.

---
##### ✅ B. Upsize the worker nodes and activate autoshuffle partitions

**Correct.**

- Larger worker nodes provide more executor memory and CPU.
- Auto-optimized shuffle partitions (Databricks feature) adjusts the number of shuffle partitions dynamically, helping avoid partitions that are too large and reducing the risk of executor OOM errors.

---

##### Why the other options are incorrect

---

##### ❌ C. Upsize the driver node and deactivate autoshuffle partitions

- Driver memory is mainly used for query planning and collecting results.
- Most OutOfMemoryError issues during ETL occur on executors (workers), not the driver.
- Disabling auto shuffle optimization can actually make shuffle performance worse.

---

##### ❌ D. Cache the dataset in order to boost the query performance

- Caching stores data in memory. If you're already running out of memory, caching may increase memory pressure and worsen the problem unless the dataset is reused and there is sufficient memory.

---

##### ❌ E. Fix the shuffle partitions to 50 to ensure the allocation

- There is no universally correct value such as 50. Too few partitions can make each partition too large and increase the chance of OOM.
- Databricks generally recommends letting Auto-Optimized Shuffle determine an appropriate number of partitions.

**Question #:11**

**A data engineer is writing a script that is meant to ingest new data from cloud storage. In the event of the Schema change, the ingestion should fail. It should fail until the changes downstream source can be found and verified as intended changes.**

**Which command will meet the requirements?**

A. failonNewColumns  
B. none  
C. rescue  
D. addNewColumns

---
**The correct answer is: ✅ A. failOnNewColumns**

---

**Explanation**

This question refers to Databricks Auto Loader and its schema evolution modes (`cloudFiles.schemaEvolutionMode`).

**The requirement is:**

- Ingest data from cloud storage.
- If the schema changes (e.g., a new column appears), the ingestion must fail.
- The pipeline should remain failed until the downstream changes are reviewed and intentionally handled.

**The schema evolution mode that satisfies this requirement is:**

`.option("cloudFiles.schemaEvolutionMode", "failOnNewColumns")`

This causes Auto Loader to:

- Detect newly added columns.
- Stop the stream/job.
- Require the engineer to review and update the schema before restarting.

---

**Schema Evolution Modes**

| Mode                | Behavior                                   | Use Case                  |
|---------------------|--------------------------------------------|---------------------------|
| ✅ failOnNewColumns | Stops ingestion when new columns are detected| Strict schema enforcement |
| addNewColumns       | Automatically evolves the schema by adding new columns | Flexible ingestion        |
| rescue              | Places unexpected columns into the rescued data column (`_rescued_data`) | Preserve unknown fields without failing |
| none                | Ignores schema evolution; new columns are not added to the schema | Fixed schema, no evolution |

---

**Why the other options are incorrect**

- ❌ **B. none**  
  Does not fail the ingestion on new columns.  
  Ignores schema evolution rather than enforcing review.

- ❌ **C. rescue**  
  Unknown columns are stored in the `_rescued_data` column.  
  The pipeline continues running.  
  Does not fail the ingestion.

- ❌ **D. addNewColumns**  
  Automatically updates the schema to include new columns.  
  This is the opposite of the requirement.

---

**Databricks Auto Loader Example**

python
df = (
    spark.readStream
         .format("cloudFiles")
         .option("cloudFiles.format", "json")
         .option("cloudFiles.schemaEvolutionMode", "failOnNewColumns")
         .load("/mnt/input")
)


With this configuration:

- If a new column is detected, the stream fails.
- After validating the schema change, you can update your downstream logic and restart the pipeline.

---

**Exam Tip (Databricks Certified Data Engineer Associate)**

Remember the four schema evolution modes:

- ✅ failOnNewColumns → Stop the pipeline on schema changes.
- addNewColumns → Automatically evolve the schema.
- rescue → Store unexpected fields in `_rescued_data`.
- none → Ignore schema evolution.

**Question #:12**

**A data engineer needs to ingest from both streaming and batch sources for a firm that relies on highly accurate data. Occasionally, some of the data picked up by the sensors that provide a streaming input are outside the expected parameters. If this occurs, the data must be dropped, but the stream should not fail.**

**Which feature of Delta Live Tables meets this requirement?**

A. Change Data Capture  
B. Error Handling  
C. Monitoring  
D. Expectations


---

**The correct answer is: ✅ D. Expectations**

---

**Explanation**

Delta Live Tables (DLT) Expectations are data quality rules that validate incoming data during pipeline execution.

---

**The question states:**

- Data comes from streaming and batch sources.
- Some sensor readings are outside expected parameters (invalid).
- Invalid records should be dropped.
- The stream should continue running (should not fail).

This is exactly what DLT Expectations are designed for.

---

**For example:**

python
@dlt.expect_or_drop("valid_temperature", "temperature BETWEEN -40 AND 100")


- Valid records are processed.
- Invalid records are dropped.
- The pipeline continues running without failure.

---

**Why the other options are incorrect**

- ❌ **A. Change Data Capture (CDC)**  
  Used to process inserts, updates, and deletes from source systems.  
  Not intended for validating or dropping bad sensor data.

- ❌ **B. Error Handling**  
  DLT doesn't have a specific feature called Error Handling for this purpose.  
  Data quality rules are implemented using Expectations.

- ❌ **C. Monitoring**  
  Monitoring provides pipeline metrics, logs, and observability.  
  It does not filter or drop invalid records.

---

**Delta Live Tables Expectations**

| Expectation         | Behavior                                      |
|---------------------|-----------------------------------------------|
| @dlt.expect()       | Records violations but continues processing.   |
| @dlt.expect_or_drop() | Drops invalid records and continues processing. ✅ |
| @dlt.expect_or_fail() | Fails the pipeline when invalid records are found. |

---

**Example**

python
import dlt

@dlt.table
@dlt.expect_or_drop("valid_sensor_value", "sensor_value BETWEEN 0 AND 100")
def sensor_data():
    return spark.readStream.table("bronze_sensor")


If `sensor_value = 120`:

- ❌ Record is dropped.
- ✅ Pipeline continues processing.

---

**Exam Tip (Databricks Certified Data Engineer Associate)**

Remember these three DLT expectation behaviors:

- ✅ expect → Keep records, log violations.
- ✅ expect_or_drop → Drop bad records, continue processing.
- ✅ expect_or_fail → Fail the pipeline when bad data is found.

---

**Explanation**

The OPTIMIZE command in Delta Lake compacts small files into larger files to improve query performance.

If you run:


OPTIMIZE sales;


and then immediately run:


OPTIMIZE sales;


again without any new data being added, the second execution has no effect because there are no additional small files to compact.

This behavior is called **idempotency**—running the same operation multiple times on unchanged data produces the same result.

---

**Why the other options are incorrect**

---

**✅ A. It has no effect because it is idempotent.**

Correct.

OPTIMIZE only rewrites files when there is something to optimize.
If the table hasn't changed since the last optimization, the second run does nothing.

---

**❌ B. It changes the number of tuples per file significantly.**

OPTIMIZE reorganizes files, not the logical number of rows (tuples).
A second run on unchanged data does not continue modifying file layout.

---

**❌ C. It further reduces file sizes by re-clustering the data.**

Once the files are already optimally compacted, another OPTIMIZE run does not further reduce file sizes.
There is nothing left to optimize.

---

**❌ D. It triggers a full liquid clustering.**

OPTIMIZE does not automatically trigger a full liquid clustering operation.
For tables using Liquid Clustering, OPTIMIZE incrementally reclusters data as needed, but it is not a full reclustering every time it runs.

sql

)

LOCATION 'abfss://container@storageaccount.dfs.core.windows.net/sales/';


The data remains in ADLS, and Databricks only stores the table metadata.
