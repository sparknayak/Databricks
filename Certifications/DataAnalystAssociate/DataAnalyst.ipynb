**Question #6:**  
**A data analyst has set up a SQL query to run every four hours on a SQL endpoint, but the SQL endpoint is taking too long to start up with each run. Which of the following changes can the data analyst make to reduce the start-up time for the endpoint while managing costs?**

- A. Reduce the SQL endpoint cluster size  
- B. Increase the SQL endpoint cluster size  
- C. Turn off the Auto stop feature  
- D. Increase the minimum scaling value  
- E. Use a Serverless SQL endpoint


**The correct answer is E. Use a Serverless SQL endpoint.**

---

###### Why Option E is Correct

- **Rapid Startup:** Serverless SQL endpoints (now often called Serverless SQL warehouses) offer nearly instant startup times, typically between 2 and 6 seconds. In contrast, classic SQL endpoints often take several minutes (approx. 4 minutes) to provision resources from the cloud provider.
- **Cost Management:** Because serverless endpoints start so quickly, you can set a very short auto-stop threshold (as low as 1 minute via API). This ensures that once the query finishes its four-hourly run, the compute resources are shut down almost immediately, preventing idle time costs.
- **Fully Managed:** Databricks manages the compute pool, so users do not need to manually configure or "prewarm" clusters to achieve high performance.

---

###### Why Other Options are Incorrect

- ❌ **A & B (Cluster Size):** Changing the size of the cluster primarily impacts query execution speed (parallelism), not the initial startup time. Larger clusters may actually take longer to start if the cloud provider has difficulty fulfilling the larger instance request.
- ❌ **C (Turn off Auto stop):** This would keep the endpoint running 24/7. While this eliminates startup delays for the four-hourly runs, it is not cost-effective, as you would be paying for idle compute during the 3.5+ hours between each run.
- ❌ **D (Increase minimum scaling):** Increasing the minimum scaling value ensures a certain number of clusters are always running within an active endpoint. However, if the entire endpoint is stopped (due to auto-stop), this setting does not prevent the initial startup delay of the endpoint itself.

**Question #30**

A data analyst is processing a complex aggregation on a table with zero null values and their query returns the following result:

| group_1 | group_2 | sum |
|---------|---------|-----|
| null    | null    | 100 |
| null    | Y       | 70  |
| null    | Z       | 30  |
| A       | null    | 50  |
| A       | Y       | 30  |
| A       | Z       | 20  |
| B       | null    | 50  |
| B       | Y       | 40  |
| B       | Z       | 10  |

Which of the following queries did the analyst run to obtain the above result?

**A.**
```sql
SELECT
    group_1,
    group_2,
    count(values) AS count
FROM my_table
GROUP BY group_1, group_2 INCLUDING NULL;
```

**B.**
```sql
SELECT
    group_1,
    group_2,
    count(values) AS count
FROM my_table
GROUP BY group_1, group_2 WITH ROLLUP;
```

**C.**
```sql
SELECT
    group_1,
    group_2,
    count(values) AS count
FROM my_table
GROUP BY group_1, group_2;
```

**D.**
```sql
SELECT
    group_1,
    group_2,
    count(values) AS count
FROM my_table
GROUP BY group_1, group_2, (group_1, group_2);
```

**E.**
```sql
SELECT
    group_1,
    group_2,
    count(values) AS count
FROM my_table
GROUP BY group_1, group_2 WITH CUBE;
```


**The correct answer is E. WITH CUBE** ✅

---

## Why?

The result contains:

- Individual combinations: (A,Y), (A,Z), (B,Y), (B,Z)
- Subtotals for group_1: (A,NULL), (B,NULL)
- Subtotals for group_2: (NULL,Y), (NULL,Z)
- Grand total: (NULL,NULL)

This is exactly what `GROUP BY ... WITH CUBE` produces.

```sql
SELECT group_1, group_2, SUM(values) AS sum
FROM my_table
GROUP BY group_1, group_2
WITH CUBE;
```

---

## Understanding the result

| group_1 | group_2 | sum | Meaning |
|---------|---------|-----|----------|
| NULL | NULL | 100 | Grand total |
| NULL | Y | 70 | Total for Y |
| NULL | Z | 30 | Total for Z |
| A | NULL | 50 | Total for A |
| A | Y | 30 | A + Y |
| A | Z | 20 | A + Z |
| B | NULL | 50 | Total for B |
| B | Y | 40 | B + Y |
| B | Z | 10 | B + Z |

---

## ROLLUP vs CUBE

This is an important exam distinction:

- **ROLLUP** → hierarchical subtotals + grand total
- **CUBE** → all possible combinations of subtotals + grand total

For two grouping columns, CUBE produces:

- (group_1, group_2)
- (group_1)
- (group_2)
- ()

That's why the result contains both (A, NULL) and (NULL, Y).

---

🎯 **Final Answer: E. GROUP BY group_1, group_2 WITH CUBE**

CREATE OR REPLACE TABLE ensures that if the table already exists, it is replaced, resulting in an empty table with the specified schema.

Why the others are wrong?

A ❌

CREATE TABLE ... USING (...)

USING is for specifying the data source/format such as DELTA, not for defining columns.

B ❌

CREATE OR REPLACE TABLE ... USING (width INT, ...)

Same issue: USING cannot be used to define the column schema.

C ✅

CREATE OR REPLACE TABLE database_name.table_name (width INT, length INT, height INT);

Correct syntax and satisfies all requirements.

D ❌

CREATE OR REPLACE TABLE table_name FROM database_name ...

Invalid SQL syntax.

🔑 Exam shortcut

If you see:

"Recreate the table, regardless of whether it already exists"

Think:

CREATE OR REPLACE TABLE

And if a specific schema is given:

CREATE OR REPLACE TABLE db.table (column definitions)
