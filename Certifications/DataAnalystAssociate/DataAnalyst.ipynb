
**Question #1:**  
**Which of the following layers of the medallion architecture is most commonly used by data analysts?**

- A. None of these layers are used by data analysts  
- B. Gold  
- C. All of these layers are used equally by data analysts  
- D. Silver  
- E. Bronze

**Correct Answer:** B. Gold

**Explanation:**

In the Medallion Architecture, data is organized into three layers:

- **Bronze Layer**
  - Raw ingested data
  - Contains uncleaned and unvalidated data
  - Mostly used by data engineers

- **Silver Layer**
  - Cleaned, filtered, and transformed data
  - Used for intermediate processing and feature engineering

- **Gold Layer**
  - Business-level aggregated and curated data
  - Optimized for reporting, dashboards, BI, and analytics
  - Most commonly used by data analysts and business users

Therefore, data analysts usually work with the Gold layer because it contains ready-to-use analytical datasets.

**Question #6:**  
**A data analyst has set up a SQL query to run every four hours on a SQL endpoint, but the SQL endpoint is taking too long to start up with each run. Which of the following changes can the data analyst make to reduce the start-up time for the endpoint while managing costs?**

- A. Reduce the SQL endpoint cluster size  
- B. Increase the SQL endpoint cluster size  
- C. Turn off the Auto stop feature  
- D. Increase the minimum scaling value  
- E. Use a Serverless SQL endpoint


**The correct answer is E. Use a Serverless SQL endpoint.**

---

###### Why Option E is Correct

- **Rapid Startup:** Serverless SQL endpoints (now often called Serverless SQL warehouses) offer nearly instant startup times, typically between 2 and 6 seconds. In contrast, classic SQL endpoints often take several minutes (approx. 4 minutes) to provision resources from the cloud provider.
- **Cost Management:** Because serverless endpoints start so quickly, you can set a very short auto-stop threshold (as low as 1 minute via API). This ensures that once the query finishes its four-hourly run, the compute resources are shut down almost immediately, preventing idle time costs.
- **Fully Managed:** Databricks manages the compute pool, so users do not need to manually configure or "prewarm" clusters to achieve high performance.

---

###### Why Other Options are Incorrect

- ❌ **A & B (Cluster Size):** Changing the size of the cluster primarily impacts query execution speed (parallelism), not the initial startup time. Larger clusters may actually take longer to start if the cloud provider has difficulty fulfilling the larger instance request.
- ❌ **C (Turn off Auto stop):** This would keep the endpoint running 24/7. While this eliminates startup delays for the four-hourly runs, it is not cost-effective, as you would be paying for idle compute during the 3.5+ hours between each run.
- ❌ **D (Increase minimum scaling):** Increasing the minimum scaling value ensures a certain number of clusters are always running within an active endpoint. However, if the entire endpoint is stopped (due to auto-stop), this setting does not prevent the initial startup delay of the endpoint itself.

**Question #8:**  
**Which of the following approaches can be used to ingest data directly from cloud-based object storage?**

- A. Create an external table while specifying the DBFS storage path to FROM  
- B. Create an external table while specifying the DBFS storage path to PATH  
- C. It is not possible to directly ingest data from cloud-based object storage  
- D. Create an external table while specifying the object storage path to FROM  
- E. Create an external table while specifying the object storage path to LOCATION

**Correct Answer:** E. Create an external table while specifying the object storage path to LOCATION

**Explanation:**

In Databricks SQL and Spark SQL, external tables can directly reference files stored in cloud object storage such as:

- Amazon Web Services S3
- Microsoft Azure Data Lake Storage
- Google Cloud Storage

This is done using the `LOCATION` clause.

**Example:**
```
CREATE EXTERNAL TABLE sales_data
USING DELTA
LOCATION 's3://bucket/path/';
```

**Why others are incorrect:**

- `FROM` is not used for specifying storage paths in table creation
- `PATH` is not the correct SQL clause here
- DBFS paths are different from directly referencing cloud object storage
- Direct ingestion from object storage is absolutely supported

✅ **Answer:** E. Create an external table while specifying the object storage path to LOCATION
