---

**Key exam concept:**

- Need real-time data → Feature Serving

##### Question #: 6

A Generative AI Engineer is creating an LLM-powered application that will need access to up-to-date news articles and stock prices.

The design requires the use of stock prices which are stored in Delta tables and finding the latest relevant news articles by searching the internet.

**How should the Generative AI Engineer architect the LLM system?**

- **A. Use an LLM to summarize the latest news articles and lookup stock tickers from the summaries to find stock prices.**

- **B. Query the Delta table for volatile stock prices and use an LLM to generate a search query to investigate potential causes of the stock volatility.**

- **C. Download and store news articles and stock price information in a vector store. Use RAG architecture to retrieve and generate information.**

- **D. Create an agent with tools for SQL querying of Delta tables and web searching, provide retrieved values to an LLM for generation of response.**

---


**The correct answer is:**


##### ✅ D. Create an agent with tools for SQL querying of Delta tables and web searching, provide retrieved values to an LLM for generation of response.

##### Why D is correct

This is a classic **AI Agent + Tool Calling** architecture.

The application needs two external, dynamic data sources:

- **Stock prices** → stored in Delta tables (structured data)
- **Latest news articles** → retrieved via web search (real-time information)

An LLM by itself does not have access to these sources. Instead, an agent can:

1. Decide when to query the Delta table using a SQL tool.
2. Decide when to search the web for recent news.
3. Retrieve both results.
4. Pass the retrieved information as context to the LLM.
5. Generate a final, accurate response.

**This approach is ideal because:**

- It accesses live data instead of relying on stale training data.
- It avoids continuously re-indexing rapidly changing information.
- It leverages the strengths of both structured querying (SQL) and unstructured retrieval (web search).

---

##### Why the other options are incorrect

**❌ A. Use an LLM to summarize the latest news articles and lookup stock tickers from the summaries to find stock prices.**

- Assumes the LLM already has access to the latest news.
- Doesn't explain how the news is retrieved.
- Not an agent architecture.

**❌ B. Query the Delta table for volatile stock prices and use an LLM to generate a search query to investigate potential causes of the stock volatility.**

- Focuses only on volatile stocks.
- Doesn't satisfy the requirement to generally answer questions using both stock prices and current news.
- Too specific and incomplete.

**❌ C. Download and store news articles and stock price information in a vector store. Use RAG architecture to retrieve and generate information.**

- RAG works well for relatively static knowledge.
- Stock prices and news change constantly, making a vector store expensive and quickly outdated.
- Real-time web search and SQL queries are more appropriate than continuously re-indexing fresh data.


##### Exam Tip (Databricks Generative AI)

Use this rule of thumb:

- **Structured data (Delta tables)** → SQL Tool
- **Real-time external information** → Web Search Tool
- **LLM needs to use external tools** → Agent
- **Static documents/knowledge base** → RAG


**✅ Correct Answer: D**

- **B.** Use a Chain-of-Thought (CoT) prompting technique to guide the LLM through a series of reasoning steps, then manually input the results from external tools for the final answer.

- **C.** Implement a framework like ReAct, which allows the LLM to generate reasoning traces and perform task-specific actions that leverage external tools if necessary.

- **D.** Encourage the LLM to make multiple API calls in sequence without planning or structuring the calls, allowing the LLM to decide when and how to use external tools spontaneously.


**The correct answer is:**

---

##### ✅ C. Implement a framework like ReAct, which allows the LLM to generate reasoning traces and perform task-specific actions that leverage external tools if necessary.

---

##### Explanation

The question asks for an LLM pipeline that can:

- Perform multi-stage reasoning
- Plan and adapt actions
- Use external tools during reasoning

The **ReAct (Reason + Act)** framework is specifically designed for this purpose.

---

##### Why C is correct

ReAct combines:

- **Reasoning** – The LLM thinks through the problem.
- **Acting** – The LLM decides when to call external tools (e.g., search engines, databases, calculators, APIs).
- **Observation** – It processes the tool's output.
- **Iteration** – It continues reasoning based on the new information until it reaches the final answer.

This makes ReAct ideal for complex tasks that require planning, decision-making, and tool use.

---

##### Why the other options are incorrect

**❌ A.**

Train the LLM to generate a single, comprehensive response without interacting with any external tools, relying solely on its pre-trained knowledge.

- Does not use external tools.
- Cannot access real-time or external information.
- Does not support adaptive planning.

**❌ B.**

Use a Chain-of-Thought (CoT) prompting technique to guide the LLM through a series of reasoning steps, then manually input the results from external tools for the final answer.

- Chain-of-Thought improves reasoning but does not enable autonomous tool usage.
- Requires manual intervention.
- Not suitable for automated multi-stage workflows.

**❌ D.**

Encourage the LLM to make multiple API calls in sequence without planning or structuring the calls, allowing the LLM to decide when and how to use external tools spontaneously.

- API calls should be planned and structured.
- Random or unstructured tool usage is inefficient and unreliable.
- Lacks the reasoning framework provided by ReAct.

---

A Generative AI Engineer at a home appliance company has been asked to design an LLM based application that accomplishes the following business objective: answer customer questions on home appliances using the associated instruction manuals.

**Which set of high-level tasks should the Generative AI Engineer's system perform?**

- **A.** Split instruction manuals into chunks and embed into a vector store. Use the question to retrieve best matched chunks of manual, and use the LLM to generate a response to the user based upon the manual retrieved.

- **B.** Create an interaction matrix of historical user questions and appliance instruction manuals. Use ALS to factorize the matrix and create embeddings. Calculate the embeddings of new queries and use them to find the best manual. Use an LLM to generate a response to the question based upon the manual retrieved.

- **C.** Calculate averaged embeddings for each instruction manual, compare embeddings to user query to find the best manual. Pass the best manual with user query into an LLM with a large context window to generate a response to the employee.

- **D.** Use an LLM to summarize all of the instruction manuals. Provide summaries of each manual and user query into an LLM with a large context window to generate a response to the user.

**The correct answer is:**

---

##### ✅ A. Split instruction manuals into chunks and embed into a vector store. Use the question to retrieve best matched chunks of manual, and use the LLM to generate a response to the user based upon the manual retrieved.

---

### Explanation

This is the standard **Retrieval-Augmented Generation (RAG)** architecture, which is the recommended approach for question answering over large document collections such as instruction manuals.

---

#### Why A is correct

The workflow follows these steps:

1. **Chunk the instruction manuals**  
   Break large manuals into smaller, meaningful sections.

2. **Generate embeddings**  
   Convert each chunk into vector embeddings.

3. **Store in a vector database**  
   Save the embeddings in a vector store for semantic search.

4. **Embed the user's question**  
   Convert the user's query into an embedding.

5. **Retrieve relevant chunks**  
   Perform vector similarity search to find the most relevant manual sections.

6. **Generate the answer**  
   Pass the retrieved chunks and the user's question to the LLM to generate an accurate response.

This approach is scalable, efficient, and minimizes hallucinations by grounding responses in the source manuals.

---

#### Why the other options are incorrect

**❌ B.**  
Create an interaction matrix of historical user questions and appliance instruction manuals. Use ALS to factorize the matrix...

- ALS (Alternating Least Squares) is a collaborative filtering algorithm used in recommendation systems.
- It is not suitable for document retrieval or semantic search in RAG systems.

**❌ C.**  
Calculate averaged embeddings for each instruction manual...

- Averaging embeddings for an entire manual loses important contextual information.
- Retrieving the whole manual is inefficient and may exceed the LLM's context window.
- RAG retrieves relevant chunks, not entire documents.

**❌ D.**  
Summarize all instruction manuals...

- Summaries can omit critical details needed to answer specific questions.
- This approach reduces accuracy and is not how RAG systems are designed.

---

#### Exam Tip

For document-based question answering, remember the standard RAG pipeline:

**Documents → Chunk → Embed → Vector Store → User Query → Embed Query → Vector Search → Retrieved Chunks + Query → LLM → Answer**

---

#### Common use cases for RAG

- Product manuals
- Company knowledge bases
- Policies and procedures
- PDFs and documentation
- Customer support systems
